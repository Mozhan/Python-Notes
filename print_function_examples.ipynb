Hi, below are a few simple examples of how to use the print function in Python.

```python
# Generate simple outputs with the print function.

# print a text. Single or double quotation marks could be used for a string.
print("Hi!")
print ('How is it going?')

# sum of 2 and 2
print(2+2)
# 2 to the power of 3
print(2**3)
# 3 multiply by 3
print(3*3)
# 3 divided by 3
print (3/3)

# Combining a string and a number to print
# There are other ways of doing this in Python too. Source: https://stackoverflow.com/questions/12018992/print-combining-strings-and-numbers
print ("My age is: ", 2)
```
