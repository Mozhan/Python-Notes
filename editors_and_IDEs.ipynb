**To install Python:**  
You can download the Python installer from https://www.python.org/downloads/

There are several different editors and IDEs that can be used to develop code in Python, including:

**PyCharm:**  
Created by Jetbrains, specifically for coding in Python. There is a free Community edition, and there is also a paid Professional edition of PyCharm.

**VS Code:**  
Lightweight source code editor, made by Microsoft. To use it for Python development, in addition to installing **Python** separately, a Python extention on the editor should also be installed. This extention provides additional support for developing code in Python.

**JupyterLab:**  
Requires local installation. Can be used via a web browser for exploratory data science.

**Google Colab:**  
You can use a browser to access this cloud service at https://colab.research.google.com/. No initial set up is needed. You can store the files on Google Drive and Github.
