```python
age = 2 # In this case, age has the integer type.

# print("My age is " + age)
# The line above causes a TypeError because an int cannot be concatenated with a string
# Therefore, to make it work, we can convert the type of the age variable from integer to string. We can use the str function for this.
# Below is an example of an explicit type conversion.

print ("My age is " + str(age))

# Below is an example of an implicit type conversion.
print (2 + 1.5)
```
