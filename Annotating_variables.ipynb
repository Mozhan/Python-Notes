# Annotating variables by type

When using functions or doing object-oriented programming, it becomes important to annotate the code since we deal with more than just built-in types.
At the same time, since annotations introduce overhead, it is important not to overuse them.
An example of a direct annonation follows:

> name: str = "Joe"


It is also possible to use type comments to indirectly annotate a variable, to specify a type while avoiding overhead. Below is an example:



> name = "Joe" # type: str


In this case, the line interpreter or IDE will ignore the type, at the same time a developer can read and see which type is used.
